```python
import json
from selenium import webdriver
from selenium.webdriver.common.by import By
import time
import re

# Initialize the WebDriver
driver = webdriver.Chrome()

# Open the YouTube channel page
channel_url = "https://www.youtube.com/@kurzgesagt/videos"
driver.get(channel_url)

# Allow the page to load
time.sleep(5)

# Scroll down to load more videos (you can adjust this as needed)
scroll_pause_time = 2
last_height = driver.execute_script("return document.documentElement.scrollHeight")
while True:
    driver.execute_script("window.scrollTo(0, document.documentElement.scrollHeight);")
    time.sleep(scroll_pause_time)
    new_height = driver.execute_script("return document.documentElement.scrollHeight")
    if new_height == last_height:
        break
    last_height = new_height

# Now retrieve the video titles and store them in the list
video_names = []

# Re-fetch the list of video elements after scrolling
video_elements = driver.find_elements(By.ID, 'video-title')

for a in video_elements:
    video_names.append(a.text)

# Initialize a dictionary to store metadata
metadata = {}

# Iterate over the video elements, click each video, and extract metadata
for i in range(len(video_names)):
    try:
        # Re-fetch video elements before each interaction to avoid stale reference
        video_elements = driver.find_elements(By.ID, 'video-title')

        # Scroll the video into view before clicking
        driver.execute_script("arguments[0].scrollIntoView(true);", video_elements[i])
        time.sleep(1)

        # Click on the video element
        driver.execute_script("arguments[0].click();", video_elements[i])
        time.sleep(3)  # Wait for the video page to load

        # Extract metadata for the video
        container = driver.find_element(By.ID, "info-container")
        driver.execute_script("arguments[0].click();", container)
        
        meta_text_str = container.text  # Extract metadata from the container

        # Use regex to extract views, date, and products from the metadata text
        views = re.search(r'([\d,]+ views)', meta_text_str)
        date = re.search(r'(\d{1,2} \w+ \d{4})', meta_text_str)
        products = re.search(r'(\d+ products)', meta_text_str)

        # Store the extracted data in the dictionary
        metadata[f'video_{i}'] = {
            "video title": video_names[i],
            "views": views.group(1) if views else "N/A",
            "date": date.group(1) if date else "N/A",
            "products": products.group(1) if products else "N/A"
        }

        print(metadata[f'video_{i}'])  # Output the metadata for this video

        # Navigate back to the videos page
        driver.back()
        time.sleep(3)  # Wait for the videos page to reload

    except Exception as e:
        print(f"An error occurred for video {i}: {e}")
        driver.back()
        time.sleep(2)  # Wait for the page to load

# Save metadata to a JSON file
with open('metadata.json', 'w') as json_file:
    json.dump(metadata, json_file, indent=4)

print("Metadata saved to metadata.json")

# Close the browser
driver.quit()
```

```python
def scroll_to_load_all_videos(driver, scroll_pause_time=2):
    """
    Scrolls down the YouTube page to load more videos.
    
    Args:
        driver: Selenium WebDriver object used for browsing.
        scroll_pause_time: Time to pause between scrolls (default is 2 seconds).
    """
    last_height = driver.execute_script("return document.documentElement.scrollHeight")
    
    while True:
        # Scroll down to the bottom of the page
        driver.execute_script("window.scrollTo(0, document.documentElement.scrollHeight);")
        
        # Wait for the page to load more content
        time.sleep(scroll_pause_time)
        
        # Calculate new scroll height and compare with last scroll height
        new_height = driver.execute_script("return document.documentElement.scrollHeight")
        if new_height == last_height:
            break  # Exit the loop if no more new content is loaded
        last_height = new_height
```
